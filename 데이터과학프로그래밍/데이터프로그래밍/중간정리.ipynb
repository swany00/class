##### 벡터 정리

```r
x <- 1 ; x = 1 # 3개 모두 같은 효과이다.
```

```r
x -> 1 # 1은 숫자라 변수할당이 안된다.
```

```r
x -> y # 변수끼리는 할당이 가능하다.
```

```r
t0 <- c(5,4,3,4,5,6) #벡터를 생성할때는 c()함수를 사용한다. 
t <- list(5, 4, 3, 4, 5, 6) #리스트 (list): 서로 다른 데이터 타입을 포함할 수 있는 구조
t <- data.frame(values = c(5, 4, 3, 4, 5, 6)) #데이터 프레임 (data.frame): 행과 열로 구성된 테이블 형태로, 
                                              #각 열은 서로 다른 데이터 타입을 가질 수 있다.
t <- matrix(c(5, 4, 3, 4, 5, 6), nrow = 2) #배열 (array): 다차원 데이터를 저장한 구조 #모든타입동일
t <- array(c(5, 4, 3, 4, 5, 6), dim = c(2, 3)) #행렬 (matrix): 동일한 데이터 타입으로 구성된 2차원 배열
```
