##**과제5**##

$2*k \ (k=1, 2, ..., 300)$으로 $100 × 3$행렬을 만들고, 각 열별로 3으로 나누었을 때 0이 되는 값들을 추출하는 R 코드를 작성하여 그 코드와 결과를 제출하시오.


- 행렬간의 산술연산

  - 행렬 A와 B의 합(차)은 두 행렬이 같은 형태일 때 정의된다.

  - 같은 위치의 원소끼리의 합(차)로 정의된다.**bold text**
	\begin{eqnarray*}
		& A = \left(a_{ij}\right)_{i=1,\ldots,n; j=1, \ldots, p}, ~ B = \left(b_{ij}\right)_{i=1,\ldots,n;j=1,\ldots,p} & \\
		& A \pm B = \left(a_{ij} \pm b_{ij}\right)_{i=1,\ldots,n;j=1,\ldots,p}&
	\end{eqnarray*}

```r
A = matrix(1:4, nrow = 2, ncol = 2)
B = matrix(5:8, nrow = 2, ncol = 2)
print(A); print(B); print(A+B)
```

```r
print(A+B)
```

```r
print(A-B)
```

 - 행렬 A와 B의 곱은 A의 열 수와 B의 행 수가 같을 때 정의된다
	\begin{eqnarray*}
		& A = \left(a_{ij}\right)_{i=1,\ldots,n; j=1, \ldots, p}, ~ B = \left(b_{jk}\right)_{j=1,\ldots,p;k=1,\ldots,q} & \\
		& AB = \left(\sum_{j=1}^pa_{ij}b_{jk}\right)_{i=1,\ldots,n;k=1,\ldots,q}&
	\end{eqnarray*}

```r
A = matrix(1:6, nrow = 3, ncol = 2)
B = matrix(7:12, nrow = 2, ncol = 3)
print(A %*% B)
```

- 행렬 A와 스칼라 c의 곱은 다음과 같이 정의된다.
	\begin{eqnarray*}
		& A = \left(a_{ij}\right)_{i=1,\ldots,n; j=1, \ldots, p}, ~ cA =  \left(ca_{ij}\right)_{i=1,\ldots,n; j=1, \ldots, p}&
	\end{eqnarray*}

```r
print(A*3)
```

- 행렬 A의 전치(transpose) 행렬은 행과 열을 바꾼 행렬이다.
	\begin{eqnarray*}
		& A = \left(a_{ij}\right)_{i=1,\ldots,n; j=1, \ldots, p}, ~ A^\top =  \left(a_{ji}\right)_{j=1,\ldots,p; i=1, \ldots, n}&
	\end{eqnarray*}

```r
A = matrix(6:1, nrow = 3, ncol = 2)
print(A)
print(t(A))
```


- 정사각행렬 $C$의 역행렬 $C^{−1}$은 다음과 같이 정의된다.
$$C^{-1}C=CC^{-1}=I_n,~I_n\mbox{ : 항등행렬}$$

```r
C = matrix(c(16, 4, 1, 4, 4, 1, 1, 1, 1), 3, 3)
print(C)
print(solve(C))
```

```r
print(solve(C)%*%C)
```

  - 행렬의 곱에 %*%이 아니라 * 기호를 쓰면 각 요소별 연산이 수행된다.
     - 두 행렬의 크기가 같을 때 정의된다.
	\begin{eqnarray*}
		& A = \left(a_{ij}\right)_{i=1,\ldots,n; j=1, \ldots, p}, ~ B = \left(b_{ij}\right)_{i=1,\ldots,n;j=1,\ldots,p} & \\
		& A * B = \left(a_{ij}b_{ij}\right), ~ A / B = \left(a_{ij} / b_{ij}\right)&
	\end{eqnarray*}

```r
A = matrix(6:1, nrow = 3, ncol = 2)
B = matrix(1:6, nrow = 3, ncol = 2)
print(A*B)
print(A/B)
```

선형연립방정식
-- 행렬을 이용하여 다음 선형연립방정식의 해를 구하여라.
\begin{eqnarray*}
		2x + 3y &=& 5 \\
		3x + 4y &=& 4 \\
		\left(\begin{array}{cc} 2 & 3 \\ 3 & 4\end{array}\right)\left(\begin{array}{c}x \\ y\end{array}\right) &=& \left(\begin{array}{cc}5 \\ 4\end{array}\right)
\end{eqnarray*}
아래 관계를 이용하여 문제를 푸시오.
\begin{eqnarray*}
		A x = b
		\rightarrow
		x = A^{-1}b
\end{eqnarray*}



```r
# 행렬 A 정의
A <- matrix(c(2, 3, 3, 4), nrow = 2, ncol = 2)
# 벡터 b 정의
b <- matrix(c(5, 4), nrow = 2, ncol = 1)
# 선형 연립방정식의 해 구하기
x <- solve(A, b)
# 결과 출력
x
print(x)
```

- 척도화 scaling [평균-표준편차 표준화]

$$ z_{ij} = \frac{x_{ij} - m_j}{s_j}, ~ i = 1, \ldots, n; ~ j = 1, \ldots, p, $$

여기서 $m_j$는 열 $j$의 평균이고, $s_j$는 열 $j$의 표준편차이다.

*이탤릭체 텍스트*

```r
n <- 100
set.seed(111)
X <- matrix(rnorm(n*4), n, 4)
print(dim(X))
m <- apply(X, 2, mean) ; print(m)
s <- apply(X, 2, sd) ; print(s)
```

```r
X.1 <- t(apply(X, 1, "-", m)) # 행의 각 원소에 대해서 다른 평균으로 뺌
X.2 <- t(apply(X.1, 1, "/", s)) # goddml 각 원소에 대해서 다른 표준편차들로 나눔
# print(X.2)
m <- apply(X.2, 2, mean) ; print(m)
s <- apply(X.2, 2, sd) ; print(s)
```

```r
print(head(round(X.2, 2), 3))  # 둘째 자리에서 반올림하고 앞의 세개의 행만 출력
a = round( apply(X.2, 2, scale), 2) # 열 단위로 표준편차가 1이 되게 하는 변환
print(apply(a,2,sd))
```

```r
ss = head(round(t((t(X) - m)/s - m ),2),3) # 열단위로 벡터화되기 때문에 전치행렬을 이용해서 표준화를 하고 다시 전치행렬을 이용해서 변환
print(ss)
```

##**과제7**##

행렬의 각 열에 대하여, 최소값 0과 최대값 1이 되도록 하는 R 코드를
작성하시오.
$$z_{ij} = \frac{x_{ij} - m_j}{M_j - m_j}, ~ i = 1, \ldots, n; ~ j = 1, \ldots, p,$$
여기서 $m_j$는 열 $j$의 최소값이고, $M_j$는 열 $j$의 최대값이다.

```r
# 샘플 데이터 생성
set.seed(123)
n <- 100  # 행 수
p <- 4    # 열 수
X <- matrix(rnorm(n * p), n, p)

# 최소값과 최대값 계산
m <- apply(X, 2, min)  # 각 열의 최소값
M <- apply(X, 2, max)  # 각 열의 최대값

# z_ij 계산
Z <- t(apply(X, 1, function(x) (x - m) / (M - m)))

# 결과 확인
print(head(Z, 3))  # 변환된 결과의 첫 3개 행 출력

min_Z <- apply(Z, 2, min)  # 각 열의 최소값
max_Z <- apply(Z, 2, max)  # 각 열의 최대값

# 최소값과 최대값 출력
min_Z
max_Z
```
