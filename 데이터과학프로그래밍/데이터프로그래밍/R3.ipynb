```r
members = list(leaders = c("park","choi"),assistants = "kang",workers = c("lee","kim","hong","son"),number=c(4,6))
print(members)
```

```r
print(class(members)); print(names(members))
length(members) ;print(mode(members))
# mode가 list를 뱉는 이유는 members를 구성하는게 다 다르기 때문
```

```r
print(members[[3]])
print(members[["leaders"]])
print(class(members[["leaders"]]))
print(mode(members[["leaders"]]))
```

```r
print(members[[c(3,4)]]) ; print()
```

```r
#범주형변수
factor는  리스트를 범주형으로 바꿔주는 역할
factor로 변환한 것을 mode로 확인하면 숫자로 보고있다.
class로 보면 factor다.

factor를 unclass하면 문자를 숫자로 바꾸는 과정이다.

factror를 쓰는 이유는 범주형이 많다
r은 범형에 강하다.
```
